```python
import matplotlib.pyplot as plt
import seaborn as sns

# Figure
fig, ax = plt.subplots(figsize=(13,9))

# Colors
colors = ["#FF595E","#FFCA3A","#8AC926","#1982C4","#6A4C93"]

# Count Plot
sns.countplot(
    x="race/ethnicity",
    data=df,
    palette=colors,
    edgecolor="black",
    linewidth=2,
    ax=ax
)

# Title
ax.set_title(
    "Race / Ethnicity Distribution of Students",
    fontsize=20,
    fontweight="bold",
    color="navy",
    pad=20
)

ax.set_xlabel("Race / Ethnicity",fontsize=14,fontweight="bold")
ax.set_ylabel("Number of Students",fontsize=14,fontweight="bold")

ax.grid(axis="y",linestyle="--",alpha=0.3)

# Values
total=len(df)

for bar in ax.patches:
    h=bar.get_height()

    ax.text(bar.get_x()+bar.get_width()/2,
            h+5,
            f"{int(h)}",
            ha="center",
            fontsize=11,
            fontweight="bold")

    ax.text(bar.get_x()+bar.get_width()/2,
            h/2,
            f"{(h/total)*100:.1f}%",
            ha="center",
            color="white",
            fontsize=11,
            fontweight="bold")

# Statistics
counts=df["race/ethnicity"].value_counts()

fig.text(
0.84,
0.74,
f"""Statistics

Group A : {counts['group A']}
Group B : {counts['group B']}
Group C : {counts['group C']}
Group D : {counts['group D']}
Group E : {counts['group E']}
""",
fontsize=10,
bbox=dict(facecolor="#FFF4CC",
          edgecolor="orange",
          boxstyle="round,pad=0.5")
)

# Insight
fig.text(
0.05,
0.14,
"""INSIGHT

• Group C has the highest
  number of students.

• Groups B and D also have
  good representation.

• Group A has the lowest
  number of students.
""",
fontsize=10,
bbox=dict(facecolor="#D6EAF8",
          edgecolor="blue",
          boxstyle="round,pad=0.5")
)

# Conclusion
fig.text(
0.63,
0.14,
"""CONCLUSION

• Students belong to all
  five ethnic groups.

• Dataset is fairly balanced
  for analysis.
""",
fontsize=10,
bbox=dict(facecolor="#D5F5E3",
          edgecolor="green",
          boxstyle="round,pad=0.5")
)

# Summary
fig.text(
0.15,
0.05,
"SUMMARY : Group C has the highest representation, while Group A has the lowest representation in the dataset.",
fontsize=11,
fontweight="bold",
bbox=dict(facecolor="#F5EEF8",
          edgecolor="purple",
          boxstyle="round,pad=0.5")
)

# Space for boxes
plt.subplots_adjust(bottom=0.32,right=0.82)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Average Scores
avg_scores = df[['math score','reading score','writing score']].mean()

subjects = ['Math','Reading','Writing']
scores = avg_scores.values

overall_avg = np.mean(scores)

# Figure
fig, ax = plt.subplots(figsize=(12,8))

# Colorful Bars
colors = ["#FF595E", "#1982C4", "#8AC926"]

bars = ax.bar(subjects,
              scores,
              color=colors,
              edgecolor="black",
              linewidth=2)

# Average Line
ax.axhline(overall_avg,
           color="red",
           linestyle="--",
           linewidth=3,
           label=f"Overall Average = {overall_avg:.2f}")

# Value Labels
for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x()+bar.get_width()/2,
            h+1,
            f"{h:.2f}",
            ha='center',
            fontsize=12,
            fontweight='bold')

# Title
ax.set_title("📊 Average Subject Scores",
             fontsize=22,
             fontweight="bold",
             color="darkblue")

ax.set_xlabel("Subjects",
              fontsize=14,
              fontweight="bold")

ax.set_ylabel("Average Marks",
              fontsize=14,
              fontweight="bold")

ax.set_ylim(0,90)

ax.grid(axis='y',
        linestyle='--',
        alpha=0.3)

ax.legend(loc="upper left")

# Statistics Box
ax.text(
2.35,
88,
f"""📈 Statistics

Math : {scores[0]:.2f}
Reading : {scores[1]:.2f}
Writing : {scores[2]:.2f}

Overall Avg : {overall_avg:.2f}
""",
fontsize=10,
verticalalignment='top',
bbox=dict(boxstyle="round",
          facecolor="#FFF9C4",
          edgecolor="orange")
)

# Insight
fig.text(
0.08,
0.08,
"""💡 INSIGHT

• Reading has the highest average marks.

• Mathematics has the lowest average marks.

• Writing performance is also very good.
""",
fontsize=10,
bbox=dict(boxstyle="round",
          facecolor="#D6EAF8",
          edgecolor="blue")
)

# Conclusion
fig.text(
0.60,
0.08,
"""✅ CONCLUSION

Students perform better in
Reading and Writing
than Mathematics.
""",
fontsize=10,
bbox=dict(boxstyle="round",
          facecolor="#D5F5E3",
          edgecolor="green")
)

# Summary
fig.text(
0.18,
0.01,
"🏆 SUMMARY : Reading has the highest average score, followed by Writing. Mathematics has the lowest average score among the three subjects.",
fontsize=11,
fontweight='bold',
bbox=dict(boxstyle="round",
          facecolor="#F5EEF8",
          edgecolor="purple")
)

# Layout
plt.subplots_adjust(bottom=0.28, right=0.82)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Correlation Matrix
corr = df[['math score','reading score','writing score']].corr()

# Figure
fig, ax = plt.subplots(figsize=(12,8))

# Heatmap
sns.heatmap(
    corr,
    annot=True,
    cmap="Spectral",
    linewidths=3,
    linecolor="white",
    square=True,
    fmt=".2f",
    annot_kws={"fontsize":14,"fontweight":"bold"},
    cbar_kws={"shrink":0.8},
    ax=ax
)

# Title
ax.set_title(
    "Correlation Heatmap of Student Scores",
    fontsize=20,
    fontweight="bold",
    color="darkblue",
    pad=20
)

# Statistics Box (Right Side)
fig.text(
0.84,
0.78,
f"""Statistics

Math - Reading : {corr.iloc[0,1]:.2f}

Math - Writing : {corr.iloc[0,2]:.2f}

Reading - Writing : {corr.iloc[1,2]:.2f}

Overall :
Strong Positive Correlation
""",
fontsize=10,
bbox=dict(facecolor="#FFF3CD",
          edgecolor="orange",
          boxstyle="round,pad=0.6")
)

# Insight
fig.text(
0.06,
0.08,
"""INSIGHT

• Reading and Writing have
  the strongest relationship.

• Students scoring high in
  Reading usually score high
  in Writing.

• Math is also positively
  correlated.
""",
fontsize=10,
bbox=dict(facecolor="#D6EAF8",
          edgecolor="blue",
          boxstyle="round,pad=0.6")
)

# Conclusion
fig.text(
0.63,
0.08,
"""CONCLUSION

• All subjects show
  positive correlation.

• Better performance in one
  subject is associated with
  better performance in others.
""",
fontsize=10,
bbox=dict(facecolor="#D5F5E3",
          edgecolor="green",
          boxstyle="round,pad=0.6")
)

# Summary
fig.text(
0.16,
0.01,
"SUMMARY : Reading and Writing have the strongest correlation (0.95), indicating excellent consistency in student performance.",
fontsize=11,
fontweight="bold",
bbox=dict(facecolor="#F5EEF8",
          edgecolor="purple",
          boxstyle="round,pad=0.6")
)

plt.subplots_adjust(bottom=0.25,right=0.82)

plt.show()
```
